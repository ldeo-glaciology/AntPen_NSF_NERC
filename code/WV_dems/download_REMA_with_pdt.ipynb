```python
import pdemtools as pdt
```

```python
bounds = [-2383047.628328867,1213809.5383127271,-2345123.2666645874,1240785.1485058172]
```

```python
dem = pdt.load.mosaic(
    dataset='rema',  # must be `arcticdem` or `rema`
    resolution=32,        # must be 2, 10, or 32
    bounds=bounds,        # (xmin, ymin, xmax, ymax) or a shapely geometry
)
```

```python
dem.plot()
```

```python
dem.rio.to_raster('/Users/jkingslake/Documents/projects/current_projects/AntPen_NSF_NERC/remote sensing/worldview/rema_mosaic/flask_rema_32m.tif')
```
