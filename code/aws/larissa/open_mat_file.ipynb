```python
import numpy as np
import pandas as pd
import scipy.io
import matplotlib.pyplot as plt
```

```python
D = scipy.io.loadmat('data_from_USAP-DC/daily_data_organized/flask.mat')

columns = [
    "timestamp",
    "wind_direction_deg",
    "wind_speed_m_s",
    "air_temperature_C",
    "relative_humidity_pct",
    "pressure_hPa",
    "wxt_supply_voltage_V",
    "gust_direction_deg",
    "gust_speed_m_s",
    "heater_voltage_V",
    "heater_temperature_C",
    "cpu_panel_voltage_V",
    "cpu_panel_current_A",
    "cpu_panel_temperature_C",
    "sun_down_V",
    "sun_up_V",
]

df = pd.DataFrame(D['flask'], columns=columns)

def matlab_dn_to_dt(dn):
    return pd.to_datetime(dn - 719529, unit='D')

df['timestamp'] = matlab_dn_to_dt(df['timestamp'])
df.set_index('timestamp', inplace=True)
df['air_temperature_C'].plot()
plt.ylabel('Air Temperature (°C)')
```
